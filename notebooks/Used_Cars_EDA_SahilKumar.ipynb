## 4. Categorical Analysis

## 5. Price Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_clean['price'], bins=50, kde=True)
plt.title('Price Distribution')
plt.xlabel('Price (USD)')
plt.ylabel('Number of Listings')
plt.tight_layout()
plt.show()
```

## 6. Relationship Analysis

```python
correlation_matrix = df_clean[['price', 'milage', 'model_year']].corr()
display(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
price_mileage = df_clean['price'].corr(df_clean['milage'])
price_year = df_clean['price'].corr(df_clean['model_year'])
year_mileage = df_clean['model_year'].corr(df_clean['milage'])

print(f'Price vs Mileage Pearson correlation: {price_mileage:.4f}')
print(f'Price vs Model Year Pearson correlation: {price_year:.4f}')
print(f'Model Year vs Mileage Pearson correlation: {year_mileage:.4f}')
```

## 7. Accident History Analysis

```python
accident_price = df_clean.groupby('accident')['price'].median().sort_values(ascending=False)
display(accident_price)
```

## 8. Outlier Analysis

```python
def outlier_summary(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower) | (df[column] > upper)]
    return Q1, Q3, IQR, lower, upper, len(outliers)

for column in ['price', 'milage']:
    print(column, outlier_summary(df_clean, column))
```

## 9. Key Analytical Insights

- Potential outliers were identified using the IQR method and retained.
